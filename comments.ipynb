# Общие сведения про сайт
- общее количество объявлений на ходится в диапазоне от 65 до 70 тысяч объявлений (на текущий момент ближе к 70)
- на странице отображается 25 объявлений
- очень удобный формат url-адреса сайта: https://cars.av.by/filter?price_usd[min]=1000&price_usd[max]=1500&page=2 - явно указываются и фильтры по стоимости, и страница, что позволяет очень просто циклом обойти все страницы
- при обходе сайт ломается на 121 странице :) причём не важно, долистал до 121 страницы вручную или вбил сразу 121

# Подготовительные мероприятия
- фильтрами по стоимости разбить все объявления на группы по ~3000 объявлений
- полученные 23 - 25 групп обойти циклом по номеру страницы и собрать все ссылки на объявления
- проверить и удалить дубликаты в списке
- обойти следующим циклом все полученные объявления и собрать данные по каждому объявлению
- сформировать датасет и сохранить его в csv файл
